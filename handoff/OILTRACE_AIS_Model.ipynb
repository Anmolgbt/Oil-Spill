```python
# Make a copy so we don't accidentally change the original dataset
ais_model = ais.copy()

# Convert time to datetime
ais_model["BaseDateTime"] = pd.to_datetime(
    ais_model["BaseDateTime"],
    errors="coerce"
)

# Sort by vessel and time
ais_model = ais_model.sort_values(
    ["MMSI", "BaseDateTime"]
).reset_index(drop=True)

print("Total AIS records:", len(ais_model))
print("Number of vessels:", ais_model["MMSI"].nunique())
print("Time range:")
print(ais_model["BaseDateTime"].min(), "to", ais_model["BaseDateTime"].max())
```

```python
# Create movement features for each vessel

ais_model["speed_change"] = (
    ais_model.groupby("MMSI")["SOG"].diff().abs()
)

ais_model["course_change"] = (
    ais_model.groupby("MMSI")["COG"].diff().abs()
)

ais_model["time_gap_minutes"] = (
    ais_model.groupby("MMSI")["BaseDateTime"]
    .diff()
    .dt.total_seconds()
    / 60
)

# Remove the first record of each vessel
# because it has no previous position to compare with
ais_model = ais_model.dropna(
    subset=["speed_change", "course_change", "time_gap_minutes"]
)

print("Records after feature creation:", len(ais_model))
display(
    ais_model[
        ["MMSI", "BaseDateTime", "LAT", "LON",
         "SOG", "COG", "speed_change",
         "course_change", "time_gap_minutes"]
    ].head(10)
)
```

```python
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

# Features used to learn vessel behaviour
features = [
    "SOG",
    "speed_change",
    "COG",
    "course_change",
    "time_gap_minutes"
]

X = ais_model[features].copy()

# Scale the features so one feature doesn't dominate because of its size
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train anomaly detection model
model = IsolationForest(
    n_estimators=200,
    contamination=0.02,
    random_state=42
)

model.fit(X_scaled)

print("Model trained successfully!")
print("Training records:", len(X))
```

```python
# Get anomaly scores for every AIS record
ais_model["anomaly_raw"] = model.decision_function(X_scaled)

# Convert to an easy 0–100 score
# 0 = very normal, 100 = very unusual
min_score = ais_model["anomaly_raw"].min()
max_score = ais_model["anomaly_raw"].max()

ais_model["anomaly_score"] = (
    100 * (max_score - ais_model["anomaly_raw"])
    / (max_score - min_score)
)

# Also mark the records the model considers anomalous
ais_model["is_anomaly"] = model.predict(X_scaled) == -1

print("Anomalous records:",
      ais_model["is_anomaly"].sum())

print("\nHighest anomaly records:")
display(
    ais_model.sort_values(
        "anomaly_score",
        ascending=False
    )[[
        "MMSI",
        "BaseDateTime",
        "LAT",
        "LON",
        "SOG",
        "speed_change",
        "COG",
        "course_change",
        "time_gap_minutes",
        "anomaly_score",
        "is_anomaly"
    ]].head(10)
)
```

```python
print("SOG statistics:")
print(ais_model["SOG"].describe())

print("\nTime-gap statistics (minutes):")
print(ais_model["time_gap_minutes"].describe())

print("\nLargest SOG values:")
display(
    ais_model.nlargest(10, "SOG")[
        ["MMSI", "BaseDateTime", "SOG", "COG",
         "speed_change", "course_change", "time_gap_minutes"]
    ]
)

print("\nLargest time gaps:")
display(
    ais_model.nlargest(10, "time_gap_minutes")[
        ["MMSI", "BaseDateTime", "SOG", "COG",
         "speed_change", "course_change", "time_gap_minutes"]
    ]
)
```

```python
# Clean clearly invalid AIS movement records

clean = ais.copy()

# Convert time
clean["BaseDateTime"] = pd.to_datetime(
    clean["BaseDateTime"],
    errors="coerce"
)

# Remove invalid basic values
clean = clean.dropna(
    subset=["MMSI", "BaseDateTime", "LAT", "LON", "SOG", "COG"]
)

# Keep valid geographic coordinates
clean = clean[
    clean["LAT"].between(-90, 90) &
    clean["LON"].between(-180, 180)
]

# Remove clearly impossible speed values
clean = clean[clean["SOG"].between(0, 40)]

# Sort by vessel and time
clean = clean.sort_values(
    ["MMSI", "BaseDateTime"]
).reset_index(drop=True)

# Calculate time gap
clean["time_gap_minutes"] = (
    clean.groupby("MMSI")["BaseDateTime"]
    .diff()
    .dt.total_seconds() / 60
)

# Keep only consecutive observations reasonably close together
clean = clean[
    clean["time_gap_minutes"].between(0, 10)
].copy()

print("Original records:", len(ais))
print("Clean records:", len(clean))
print("Removed records:", len(ais) - len(clean))
print("Vessels remaining:", clean["MMSI"].nunique())
```

```python
# Recalculate movement features on the CLEAN data

clean["speed_change"] = (
    clean.groupby("MMSI")["SOG"].diff().abs()
)

clean["course_change"] = (
    clean.groupby("MMSI")["COG"].diff().abs()
)

# Handle circular course changes:
# 359° → 1° is a 2° change, not 358°
clean["course_change"] = np.minimum(
    clean["course_change"],
    360 - clean["course_change"]
)

# Remove first observation of each vessel
clean = clean.dropna(
    subset=["speed_change", "course_change"]
).copy()

print("Final training records:", len(clean))
print("Vessels:", clean["MMSI"].nunique())

display(
    clean[
        [
            "MMSI",
            "BaseDateTime",
            "LAT",
            "LON",
            "SOG",
            "COG",
            "speed_change",
            "course_change",
            "time_gap_minutes"
        ]
    ].head(10)
)
```

```python
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

# Features describing vessel behaviour
features = [
    "SOG",
    "speed_change",
    "course_change",
    "time_gap_minutes"
]

X_clean = clean[features].copy()

# Scale features
scaler_clean = StandardScaler()
X_clean_scaled = scaler_clean.fit_transform(X_clean)

# Train anomaly detector
model_clean = IsolationForest(
    n_estimators=300,
    contamination=0.02,
    random_state=42
)

model_clean.fit(X_clean_scaled)

print("OILTRACE anomaly model trained!")
print("Training records:", len(X_clean))
print("Training vessels:", clean["MMSI"].nunique())
```

```python
# Generate anomaly scores
clean["anomaly_raw"] = model_clean.decision_function(X_clean_scaled)

# Convert to 0-100
# 0 = very normal
# 100 = very unusual
min_score = clean["anomaly_raw"].min()
max_score = clean["anomaly_raw"].max()

clean["anomaly_score"] = (
    100 * (max_score - clean["anomaly_raw"])
    / (max_score - min_score)
)

clean["is_anomaly"] = model_clean.predict(X_clean_scaled) == -1

print("Anomalous records:",
      clean["is_anomaly"].sum())

print("\nTop 10 unusual AIS observations:")

display(
    clean.sort_values(
        "anomaly_score",
        ascending=False
    )[[
        "MMSI",
        "BaseDateTime",
        "LAT",
        "LON",
        "SOG",
        "speed_change",
        "course_change",
        "time_gap_minutes",
        "anomaly_score",
        "is_anomaly"
    ]].head(10)
)
```

```python
# Convert individual AIS anomalies into one score per vessel

vessel_scores = (
    clean.groupby("MMSI")
    .agg(
        total_records=("MMSI", "size"),
        anomalous_records=("is_anomaly", "sum"),
        average_anomaly=("anomaly_score", "mean"),
        maximum_anomaly=("anomaly_score", "max")
    )
    .reset_index()
)

# Percentage of a vessel's observations that were anomalous
vessel_scores["anomaly_percentage"] = (
    vessel_scores["anomalous_records"]
    / vessel_scores["total_records"]
    * 100
)

# Sort by strongest overall anomalous behaviour
vessel_scores = vessel_scores.sort_values(
    "anomaly_percentage",
    ascending=False
)

print("Vessels analysed:", len(vessel_scores))

display(vessel_scores.head(15))
```

```python
import os
from sklearn.model_selection import train_test_split

# Find all images
image_extensions = (".jpg", ".jpeg", ".png", ".tif", ".tiff")

image_files = []

for root, dirs, files in os.walk(path):
    for file in files:
        if file.lower().endswith(image_extensions):
            image_files.append(os.path.join(root, file))

# Class_0 = 0, Class_1 = 1
labels = [
    1 if "/Class_1/" in f else 0
    for f in image_files
]

# 80% train, 20% temporary
train_files, temp_files, train_labels, temp_labels = train_test_split(
    image_files,
    labels,
    test_size=0.20,
    random_state=42,
    stratify=labels
)

# 10% validation, 10% test
val_files, test_files, val_labels, test_labels = train_test_split(
    temp_files,
    temp_labels,
    test_size=0.50,
    random_state=42,
    stratify=temp_labels
)

print("Total images     :", len(image_files))
print("Training images  :", len(train_files))
print("Validation images:", len(val_files))
print("Test images      :", len(test_files))
```

```python
import torch
from torch.utils.data import Dataset, DataLoader
from PIL import Image
from torchvision import transforms

# Convert each 400x400 SAR image into a format the model can use
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.Grayscale(num_output_channels=3),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

class SAR_Dataset(Dataset):
    def __init__(self, files, labels, transform=None):
        self.files = files
        self.labels = labels
        self.transform = transform

    def __len__(self):
        return len(self.files)

    def __getitem__(self, index):
        image = Image.open(self.files[index]).convert("L")
        label = self.labels[index]

        if self.transform:
            image = self.transform(image)

        return image, torch.tensor(label, dtype=torch.long)


train_dataset = SAR_Dataset(train_files, train_labels, transform)
val_dataset = SAR_Dataset(val_files, val_labels, transform)
test_dataset = SAR_Dataset(test_files, test_labels, transform)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)

print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Using device:", device)
```

```python
class OilSpillCNN(nn.Module):
    def __init__(self):
        super(OilSpillCNN, self).__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(128, 256, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d((1, 1))
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(128, 2)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x


model = OilSpillCNN().to(device)

print(model)
```

```python
import torch.nn as nn
import torch.optim as optim

# Loss function
criterion = nn.CrossEntropyLoss()

# Optimizer
optimizer = optim.Adam(
    model.parameters(),
    lr=0.001
)

print("Loss function:", criterion)
print("Optimizer:", optimizer)
```

```python
num_epochs = 10

train_losses = []
val_losses = []

train_accuracies = []
val_accuracies = []

best_val_accuracy = 0.0

for epoch in range(num_epochs):

    # =========================
    # TRAINING
    # =========================
    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in train_loader:

        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        running_loss += loss.item()

        _, predicted = torch.max(outputs, 1)

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    train_loss = running_loss / len(train_loader)
    train_accuracy = 100 * correct / total

    # =========================
    # VALIDATION
    # =========================
    model.eval()

    val_running_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():

        for images, labels in val_loader:

            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)

            val_running_loss += loss.item()

            _, predicted = torch.max(outputs, 1)

            val_total += labels.size(0)
            val_correct += (predicted == labels).sum().item()

    val_loss = val_running_loss / len(val_loader)
    val_accuracy = 100 * val_correct / val_total

    train_losses.append(train_loss)
    val_losses.append(val_loss)

    train_accuracies.append(train_accuracy)
    val_accuracies.append(val_accuracy)

    # Save best model
    if val_accuracy > best_val_accuracy:

        best_val_accuracy = val_accuracy

        torch.save(
            model.state_dict(),
            "best_oil_spill_cnn.pth"
        )

    print(
        f"Epoch [{epoch+1}/{num_epochs}] "
        f"Train Loss: {train_loss:.4f} "
        f"Train Acc: {train_accuracy:.2f}% "
        f"Val Loss: {val_loss:.4f} "
        f"Val Acc: {val_accuracy:.2f}%"
    )

print("\nTraining complete!")
print(f"Best validation accuracy: {best_val_accuracy:.2f}%")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
import numpy as np

# Load the best CNN weights
model.load_state_dict(
    torch.load("best_oil_spill_cnn.pth", map_location=device)
)

model.eval()

all_predictions = []
all_labels = []

with torch.no_grad():

    for images, labels in test_loader:

        images = images.to(device)
        labels = labels.to(device)

        outputs = model(images)

        _, predictions = torch.max(outputs, 1)

        all_predictions.extend(predictions.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())


# Convert to numpy arrays
all_predictions = np.array(all_predictions)
all_labels = np.array(all_labels)

# Metrics
accuracy = accuracy_score(all_labels, all_predictions)

precision = precision_score(
    all_labels,
    all_predictions,
    zero_division=0
)

recall = recall_score(
    all_labels,
    all_predictions,
    zero_division=0
)

f1 = f1_score(
    all_labels,
    all_predictions,
    zero_division=0
)

print("CNN TEST RESULTS")
print("=" * 40)

print(f"Accuracy : {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1-score : {f1:.4f}")

print("\nClassification Report:")
print(
    classification_report(
        all_labels,
        all_predictions,
        target_names=["Class 0", "Class 1"],
        zero_division=0
    )
)

print("\nConfusion Matrix:")
print(confusion_matrix(all_labels, all_predictions))
```

```python
# Simple prototype for backward source estimation

import numpy as np

# Example detected spill location
spill_lat = 28.57
spill_lon = -94.80

# Example environmental drift
# We'll replace these with real wind/current data later
drift_speed_kmh = 1.5
drift_direction_deg = 135   # direction oil moved toward

# Hindcast duration
hours_back = 24

# Convert direction to radians
theta = np.radians(drift_direction_deg)

# Distance travelled backward
distance_km = drift_speed_kmh * hours_back

# Approximate displacement
dx_km = distance_km * np.sin(theta)
dy_km = distance_km * np.cos(theta)

# Rough conversion
lat_per_km = 1 / 111
lon_per_km = 1 / (111 * np.cos(np.radians(spill_lat)))

source_lat = spill_lat - dy_km * lat_per_km
source_lon = spill_lon - dx_km * lon_per_km

print("Detected spill:")
print(f"  Latitude : {spill_lat}")
print(f"  Longitude: {spill_lon}")

print("\nEstimated source after", hours_back, "hours backward:")
print(f"  Latitude : {source_lat:.5f}")
print(f"  Longitude: {source_lon:.5f}")
```

```python
# ============================================
# CREATE AIS BEHAVIOURAL FEATURES
# ============================================

# Previous observation for each vessel
ais_model["prev_SOG"] = (
    ais_model.groupby("MMSI")["SOG"].shift(1)
)

ais_model["prev_COG"] = (
    ais_model.groupby("MMSI")["COG"].shift(1)
)

ais_model["prev_time"] = (
    ais_model.groupby("MMSI")["BaseDateTime"].shift(1)
)

# Speed change
ais_model["speed_change"] = (
    ais_model["SOG"] - ais_model["prev_SOG"]
).abs()

# Course change
course_diff = (
    ais_model["COG"] - ais_model["prev_COG"]
).abs()

# Handle 0/360 degree wrap-around
ais_model["course_change"] = np.minimum(
    course_diff,
    360 - course_diff
)

# Time gap in minutes
ais_model["time_gap_minutes"] = (
    ais_model["BaseDateTime"] - ais_model["prev_time"]
).dt.total_seconds() / 60

# Remove first record of each vessel
ais_model = ais_model.dropna(
    subset=[
        "speed_change",
        "course_change",
        "time_gap_minutes"
    ]
).copy()

print("Records after feature creation:", len(ais_model))
print("Vessels:", ais_model["MMSI"].nunique())

ais_model[
    [
        "MMSI",
        "BaseDateTime",
        "SOG",
        "COG",
        "speed_change",
        "course_change",
        "time_gap_minutes"
    ]
].head(10)
```

```python
# ============================================
# OILTRACE — TRAJECTORY ANALYSIS
# ============================================

# Look at vessel movement around the estimated source
trajectory_before = pd.Timedelta(hours=6)
trajectory_after = pd.Timedelta(hours=2)

trajectory_data = ais_model[
    (ais_model["BaseDateTime"] >= source_time - trajectory_before) &
    (ais_model["BaseDateTime"] <= source_time + trajectory_after)
].copy()

# Only analyse the candidate vessels
candidate_mmsi = vessel_ranking["MMSI"].tolist()

trajectory_data = trajectory_data[
    trajectory_data["MMSI"].isin(candidate_mmsi)
].copy()

# Calculate distance to estimated source
trajectory_data["distance_to_source_km"] = trajectory_data.apply(
    lambda row: haversine_km(
        source_lat,
        source_lon,
        row["LAT"],
        row["LON"]
    ),
    axis=1
)

# Sort by vessel and time
trajectory_data = trajectory_data.sort_values(
    ["MMSI", "BaseDateTime"]
)

# --------------------------------------------
# Calculate movement toward/away from source
# --------------------------------------------

trajectory_results = []

for mmsi, vessel in trajectory_data.groupby("MMSI"):

    vessel = vessel.sort_values("BaseDateTime").copy()

    # Distance at beginning and end of trajectory window
    start_distance = vessel["distance_to_source_km"].iloc[0]
    end_distance = vessel["distance_to_source_km"].iloc[-1]

    # Closest approach
    min_distance = vessel["distance_to_source_km"].min()

    closest_row = vessel.loc[
        vessel["distance_to_source_km"].idxmin()
    ]

    closest_time = closest_row["BaseDateTime"]

    # Did vessel approach the source?
    if end_distance < start_distance:
        movement = "Approaching source"
    else:
        movement = "Moving away / not approaching"

    # Approach score
    if start_distance > 0:
        approach_score = (
            (start_distance - min_distance)
            / start_distance
            * 100
        )
    else:
        approach_score = 0

    approach_score = np.clip(
        approach_score,
        0,
        100
    )

    trajectory_results.append({
        "MMSI": mmsi,
        "start_distance_km": start_distance,
        "closest_distance_km": min_distance,
        "end_distance_km": end_distance,
        "closest_time": closest_time,
        "trajectory_status": movement,
        "trajectory_score": approach_score
    })


trajectory_df = pd.DataFrame(trajectory_results)

# --------------------------------------------
# Combine with previous ranking
# --------------------------------------------

vessel_ranking = vessel_ranking.merge(
    trajectory_df[
        [
            "MMSI",
            "start_distance_km",
            "closest_distance_km",
            "end_distance_km",
            "closest_time",
            "trajectory_status",
            "trajectory_score"
        ]
    ],
    on="MMSI",
    how="left"
)

# --------------------------------------------
# FINAL OILTRACE SCORE
# --------------------------------------------

vessel_ranking["final_suspect_score"] = (
    0.40 * vessel_ranking["proximity_score"] +
    0.30 * vessel_ranking["trajectory_score"] +
    0.30 * vessel_ranking["behaviour_score"]
)

# Rank
vessel_ranking = vessel_ranking.sort_values(
    "final_suspect_score",
    ascending=False
).reset_index(drop=True)

print("\n========================================")
print("OILTRACE FINAL VESSEL RANKING")
print("========================================")

display(
    vessel_ranking[
        [
            "MMSI",
            "records",
            "min_distance_km",
            "proximity_score",
            "trajectory_status",
            "trajectory_score",
            "behaviour_score",
            "final_suspect_score"
        ]
    ]
)
```

```python
# ============================================
# OILTRACE — FINAL SUSPECT SCORE
# ============================================

# Make sure missing values do not break scoring
vessel_ranking["trajectory_score"] = (
    vessel_ranking["trajectory_score"]
    .fillna(0)
)

vessel_ranking["behaviour_score"] = (
    vessel_ranking["behaviour_score"]
    .fillna(0)
)

vessel_ranking["proximity_score"] = (
    vessel_ranking["proximity_score"]
    .fillna(0)
)

# Final weighted score
#
# Proximity   = 40%
# Trajectory  = 30%
# Behaviour   = 30%

vessel_ranking["final_suspect_score"] = (
    0.40 * vessel_ranking["proximity_score"] +
    0.30 * vessel_ranking["trajectory_score"] +
    0.30 * vessel_ranking["behaviour_score"]
)

# Rank highest first
vessel_ranking = vessel_ranking.sort_values(
    "final_suspect_score",
    ascending=False
).reset_index(drop=True)

# Add rank
vessel_ranking["rank"] = (
    vessel_ranking.index + 1
)

# Display final result
print("\n")
print("=" * 70)
print("             OILTRACE FINAL SUSPECT RANKING")
print("=" * 70)

display(
    vessel_ranking[
        [
            "rank",
            "MMSI",
            "records",
            "min_distance_km",
            "proximity_score",
            "trajectory_status",
            "trajectory_score",
            "behaviour_score",
            "final_suspect_score"
        ]
    ]
)
```

```python
# ============================================
# OILTRACE — STAGE 7: OIL SPILL FORECAST
# ============================================

import numpy as np
import pandas as pd

# --------------------------------------------
# Current detected spill location
# --------------------------------------------

spill_lat = 28.57
spill_lon = -94.80

# Estimated source from hindcast
source_lat = 28.79933
source_lon = -95.06113

print("Detected spill:")
print(f"Latitude : {spill_lat}")
print(f"Longitude: {spill_lon}")

print("\nEstimated source:")
print(f"Latitude : {source_lat}")
print(f"Longitude: {source_lon}")


# --------------------------------------------
# Simple drift assumptions
# --------------------------------------------
#
# IMPORTANT:
# These are prototype assumptions.
# They are NOT real-time ocean-current data.
#
# We will replace these later with real
# environmental data if available.

drift_speed_kmh = 1.0
drift_direction_deg = 180.0


# --------------------------------------------
# Forecast function
# --------------------------------------------

def forecast_position(
    lat,
    lon,
    speed_kmh,
    direction_deg,
    hours
):
    """
    Simple great-circle approximation for short
    ocean drift distances.
    """

    direction_rad = np.radians(direction_deg)

    distance_km = speed_kmh * hours

    # Approximate conversion
    km_per_degree_lat = 111.0
    km_per_degree_lon = 111.0 * np.cos(np.radians(lat))

    delta_lat = (
        distance_km * np.cos(direction_rad)
        / km_per_degree_lat
    )

    delta_lon = (
        distance_km * np.sin(direction_rad)
        / km_per_degree_lon
    )

    new_lat = lat + delta_lat
    new_lon = lon + delta_lon

    return new_lat, new_lon


# --------------------------------------------
# Generate forecast
# --------------------------------------------

forecast_hours = [6, 12, 24, 48]

forecast = []

for hours in forecast_hours:

    lat, lon = forecast_position(
        spill_lat,
        spill_lon,
        drift_speed_kmh,
        drift_direction_deg,
        hours
    )

    forecast.append({
        "hours_ahead": hours,
        "latitude": lat,
        "longitude": lon
    })


forecast_df = pd.DataFrame(forecast)


print("\n")
print("=" * 60)
print("        OILTRACE OIL SPILL FORECAST")
print("=" * 60)

display(forecast_df)
```

```python
# ============================================
# RECREATE OIL SPILL CNN
# ============================================

import torch
import torch.nn as nn

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)

class OilSpillCNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(128, 256, kernel_size=3, padding=1),
            nn.ReLU(),

            nn.AdaptiveAvgPool2d((1, 1))
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(128, 2)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x


cnn_model = OilSpillCNN().to(device)

print(cnn_model)
print("\nCNN recreated successfully.")
```

```python
# ============================================================
# OILTRACE — CNN VALIDATION & MODEL QUALITY
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    roc_curve,
    confusion_matrix,
    classification_report
)

print("=" * 75)
print("                 OILTRACE CNN VALIDATION")
print("=" * 75)

# ------------------------------------------------------------
# Find existing test labels / predictions
# ------------------------------------------------------------

possible_true = [
    "y_test",
    "y_true",
    "test_labels",
    "labels_test"
]

possible_pred = [
    "y_pred",
    "predictions",
    "test_predictions"
]

possible_prob = [
    "y_prob",
    "y_probs",
    "probabilities",
    "test_probabilities"
]

y_true = next(
    (globals()[x] for x in possible_true if x in globals()),
    None
)

y_pred = next(
    (globals()[x] for x in possible_pred if x in globals()),
    None
)

y_prob = next(
    (globals()[x] for x in possible_prob if x in globals()),
    None
)

# ------------------------------------------------------------
# Report what was found
# ------------------------------------------------------------

print("\nAvailable validation objects:")

print("y_true:", y_true is not None)
print("y_pred:", y_pred is not None)
print("y_prob:", y_prob is not None)

# ------------------------------------------------------------
# Basic validation
# ------------------------------------------------------------

if y_true is None or y_pred is None:

    print("\n⚠️ Existing test labels/predictions were not found.")
    print("We will retrieve them from the test dataset in the next step.")

else:

    y_true = np.asarray(y_true).astype(int)
    y_pred = np.asarray(y_pred).astype(int)

    print("\n" + "-" * 75)
    print("CLASSIFICATION METRICS")
    print("-" * 75)

    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )
    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )
    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    print(f"Accuracy  : {accuracy:.4f} ({accuracy*100:.2f}%)")
    print(f"Precision : {precision:.4f}")
    print(f"Recall    : {recall:.4f}")
    print(f"F1-score  : {f1:.4f}")

    # --------------------------------------------------------
    # Confusion matrix
    # --------------------------------------------------------

    cm_final = confusion_matrix(
        y_true,
        y_pred
    )

    print("\nConfusion Matrix:")
    print(cm_final)

    print("\nClassification Report:")
    print(
        classification_report(
            y_true,
            y_pred,
            target_names=[
                "No Oil Spill",
                "Oil Spill"
            ],
            zero_division=0
        )
    )

    # --------------------------------------------------------
    # ROC-AUC
    # --------------------------------------------------------

    if y_prob is not None:

        y_prob = np.asarray(y_prob)

        # Handle probability arrays shaped (N, 2)
        if y_prob.ndim == 2 and y_prob.shape[1] == 2:
            y_prob = y_prob[:, 1]

        auc = roc_auc_score(
            y_true,
            y_prob
        )

        print(f"ROC-AUC   : {auc:.4f}")

        # ROC curve
        fpr, tpr, thresholds = roc_curve(
            y_true,
            y_prob
        )

        plt.figure(figsize=(8, 6))

        plt.plot(
            fpr,
            tpr,
            label=f"ROC-AUC = {auc:.4f}"
        )

        plt.plot(
            [0, 1],
            [0, 1],
            linestyle="--",
            label="Random classifier"
        )

        plt.xlabel("False Positive Rate")
        plt.ylabel("True Positive Rate")
        plt.title("OILTRACE CNN ROC Curve")
        plt.legend()
        plt.grid(True)

        plt.show()

    else:

        print("\nROC-AUC could not be calculated because")
        print("probability predictions were not found.")

print("\n" + "=" * 75)
print("CNN VALIDATION STEP COMPLETE")
print("=" * 75)
```
